```python
import logging
from datetime import datetime, timezone
```

```python
job_task_timestamp_utc = datetime.now(timezone.utc)
```

```python
notebook_path = dbutils.notebook.entry_point.getDbutils().notebook().getContext().notebookPath().get()
notebook_name = notebook_path.split("/")[-1]
notebook_folder = notebook_path.split("/")[-2]
```

```python
loggerName = f"m07sparksql-{notebook_folder}-{notebook_name}"
logger = logging.getLogger(loggerName)
logger.propagate = False
logger.setLevel("INFO")

handler = logging.StreamHandler()
handler.setFormatter(logging.Formatter("%(levelname)s: %(message)s"))
logger.addHandler(handler)

job_task_timestamp_utc = datetime.now(timezone.utc)

logger.info(f"Execution Started - '{notebook_name}' notebook")
logger.info(f"Notebook Context: '{dbutils.notebook.entry_point.getDbutils().notebook().getContext().toJson()}'")
logger.info(f"job_task_timestamp_utc: '{job_task_timestamp_utc}'")
```
